## Thoughts
* If we are using p-value methods it is important that we state $\mathcal{H_i}$'s.
    * This is the **type(s)** of anomalies we are trying to detect (moderate shifts, extreme readings, persistent deviations)
    * This is how we can properly **label** our data.
        * If we don't wish to define and test them we should be using semi-supervised methods
        * We cannot properly evaluate methods without this.
            * Different packaged methods generally come with different $\mathcal{H_i}$
* How can we state minimal viable activity (MVA)?
* Consider the following parameters:
    * $l$: Trial length (seconds an simulation trial is run)
    * $h$: Sampling frequency (in hz; larger values for fine-grained simulations)
    * $r$: Number of runs (how many simulation trials are run)
* Suggested shapes for data structures:
    * Simulated Positions: $(r, m + n, l\times h+1, n_d)$
        *  m, n are the total number of sources and sensors, respectively
        * can break these up into individual position arrays
        * $n_d$ is the position dimension (2d, 3d)
        * if the positions do not vary over trials, we can have $r = 1$ here
    * Simulated distances: $(r, m, n, l\times h)$
        * ```distances[0,1,4]``` would be the distance array between source 2 and sensor 5 
    * Integrated positions: $(r, m+n, l, n_d)$
    * Integrated distances: $(r, m+n, l)$
